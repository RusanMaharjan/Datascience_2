# Fill Null values

## Filling custom records for null values

```python
df = pd.read_csv('Employees.csv')
```

```python
df.isnull().sum()
```

```python
# df.fillna(0, inplace=True)
```

```python
# df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(0)
```

```python
# df['MANAGER_ID'] = df['MANAGER_ID'].fillna(0)
# df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(0)
```

```python
df.head()
```

```python
commission = df['COMMISSION_PCT'] > 0
df[commission]
```

```python
df.isnull().sum()
```

```python
# Aggregate Functions
# ---------------------------
# mean
# max
# min
# count
# sum
```

```python
commission_mean = df['COMMISSION_PCT'].mean()
com_mean = "{:.2f}".format(commission_mean)
com_mean
# com_mean = f"{commission_mean:.2f}"
```

```python
df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(com_mean)
```

```python
df
```

```python
df.isnull().sum()
```

```python
mgr_mode = df['MANAGER_ID'].mode()
mgr_mode
```

```python
df['MANAGER_ID'] = df['MANAGER_ID'].fillna(mgr_mode)
```

```python
df.head()
```

```python
dep_mode = df['DEPARTMENT_ID'].mode()
dep_mode
```

```python
df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(dep_mode)[0]
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['HIRE_DATE'] = pd.to_datetime(df['HIRE_DATE'])
```
