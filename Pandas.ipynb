```python
# Find record of employees who works on department 60 and 90.
emp_dep = (employee['DEPARTMENT_ID'] == 60) | (employee['DEPARTMENT_ID'] == 90)
employee[emp_dep]
```

```python
# find details of employees whose salary is less than 24000, more than 7000
emp_sal = (employee['SALARY'] >= 7000) & (employee['SALARY'] <= 24000)
employee[emp_sal]
```

```python
# Find employee details who works under job AD_VP, AD_PRES, IT_PROG, AC_MGR, AC_ACCOUNT
emp_job = (employee['JOB_ID'] == 'AD_VP') | (employee['JOB_ID'] == 'AD_PRES') | (employee['JOB_ID'] == 'IT_PROG') | (employee['JOB_ID'] == 'AC_MGR') | (employee['JOB_ID'] == 'AC_ACCOUNT')
employee[emp_job]
```

```python
# between
# (employee['SALARY'] >= 7000) & (employee['SALARY'] <= 24000)
salary_between = employee['SALARY'].between(7000, 24000)
employee[salary_between]

# isin

# (employee['JOB_ID'] == 'AD_VP') | (employee['JOB_ID'] == 'AD_PRES') | (employee['JOB_ID'] == 'IT_PROG') | (employee['JOB_ID'] == 'AC_MGR') | (employee['JOB_ID'] == 'AC_ACCOUNT')
job_isin = employee['JOB_ID'].isin(['AD_VP', 'AD_PRES', 'IT_PROG', 'AC_MGR'])
employee[job_isin]
```

```python
df = pd.read_csv('HR_Dataset Refresh.csv')
df.head()
```

```python
# 1. Find details of employees with the salary range of 50000 to 70000 and who are Married.
emp_detail = ((df['Salary'] >= 50000) & (df['Salary'] <= 70000)) & (df['MaritalDesc'] == 'Married')
df[emp_detail]
```

```python
# 2. Details of employees who works under production department and their employment status is active as well.
dep_emp = (df['Department'] == 'Production       ') & (df['EmploymentStatus'] == 'Active')
df.loc[dep_emp, ['Department', 'EmploymentStatus']]
```

```python
df.columns
```

```python
# 3. Details of employees who works under IT/IS department and their absences record must be more than 8.
emp_abs = (df['Department'] == 'IT/IS') & (df['Absences'] > 8)
df.loc[emp_abs, ['Department', 'Absences']]
```
