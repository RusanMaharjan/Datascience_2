```python
# Line Chart -> Trend Analysis, Time Analysis, Sales Reporting
# Bar Chart -> Categorical Data
# Histogram / Histplot -> Skewness of data
# Scattor Plot -> Find relationship between two data
# Box Plot -> mean, median, quartile, mode, std
# Radar Chart
# Violin Plot
# Map Plot
```

```python
import matplotlib.pyplot as plt
```

## Line Chart

```python
'''
X -> Independent Variable
Y -> Dependent Variable
'''
fruits = ['Apple', 'Banana', 'Coconut', 'Orange']
sales = [150, 200, 100, 300]
revenue = [100, 150, 80, 200]
```

```python
plt.figure(figsize=(6, 4))
plt.plot(fruits, sales, c='red', marker='o', ls='dashed', label='Sales')
plt.plot(fruits, revenue, c='green', marker='*', label='Revenue')
plt.xlabel('Fruits')
plt.ylabel('Sales of Fruits')
plt.title('Total Sales of Fruits')
plt.legend()
plt.savefig('fruit_sales.png')
plt.show()
```

# Bar Plot / Bar Chart

```python
fruits = ['Apple', 'Banana', 'Coconut', 'Orange']
sales = [150, 200, 100, 300]
```

```python
plt.figure(figsize=(6, 4))
plt.bar(fruits, sales, color=['red', 'green', 'blue', 'orange'], 
        edgecolor='black', alpha=0.7)
plt.xlabel('Fruits')
plt.ylabel('Sales of Fruits')
plt.title('Total Sales of Fruits')
plt.grid(ls='--', color='black', alpha=0.2)
plt.show()
```

```python
hr = pd.read_csv("D:/Broadway/Data Science - 2/HR_Dataset Refresh.csv")
hr.head()
```

# Histogram

```python
plt.figure(figsize=(6, 4))
plt.hist(hr['Salary'], edgecolor='black')
```

```python
'''
KDE (Kernel Density Estimate) Plot
'''
```

```python
import seaborn as sns
```

```python
mean_salary = hr['Salary'].mean()
median_salary = hr['Salary'].median()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(hr['Salary'], bins=10, kde=True, color='blue')
plt.axvline(mean_salary, color='red', label='Mean')
plt.axvline(median_salary, color='green', ls='--', label='Median')
plt.legend()
```

```python
total_gender = hr.groupby('Gender')['EmpID'].count().reset_index()
total_gender
```

```python
plt.figure(figsize=(6, 4))
plt.bar(total_gender['Gender'], total_gender['EmpID'], width=0.5)
plt.xlabel('Gender')
plt.ylabel('Number of Gender')
plt.title('Total number of Gender')

for i, val in enumerate(total_gender['EmpID']):
    plt.text(total_gender['Gender'][i], val, str(val), va='bottom', ha='center')
```

```python
plt.figure(figsize=(6, 4))
ax = sns.barplot(x=total_gender['Gender'], y=total_gender['EmpID'], width=0.5,
                palette='viridis')
plt.xlabel('Gender')
plt.ylabel('Number of Gender')
plt.title('Total number of Gender')

for container in ax.containers:
    ax.bar_label(container)
```

```python
ax = sns.barplot(x=hr['PerformanceScore'], y=hr['Salary'], hue=hr['Gender'], 
            errorbar=None, palette='Blues')
plt.title('Employee Salary According to Performance Score')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')
```

```python
'''
Find total number of employees, total salary, total employee satisfaction,
total engagement survey
in each department.
'''
hr_detail = hr.groupby('Department').agg(
    Total_Employee = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Total_Emp_Sat = ('EmpSatisfaction', 'sum'),
    Total_Eng_Sur = ('EngagementSurvey', 'sum')
).reset_index()
hr_detail
```

```python
plt.figure(figsize=(6, 4))
plt.bar(hr_detail['Department'],hr_detail['Total_Employee'])
plt.xticks(rotation=10)
```

```python
plt.figure(figsize=(6, 4))
plt.bar(hr_detail['Department'],hr_detail['Total_Salary'])
plt.xticks(rotation=10)
```

```python
plt.figure(figsize=(6, 4))
plt.bar(hr_detail['Department'],hr_detail['Total_Emp_Sat'])
plt.xticks(rotation=10)
```

```python
plt.figure(figsize=(6, 4))
plt.bar(hr_detail['Department'],hr_detail['Total_Eng_Sur'])
plt.xticks(rotation=10)
```

# Subplots

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
# plt.subplots_adjust(hspace=0.5, wspace=0.5)
plt.suptitle('Department vs. Other Features') # Supplement Title
plt.subplots_adjust(hspace=0.5)

ax1 = axes[0, 0]
ax1.bar(hr_detail['Department'],hr_detail['Total_Employee'], color='blue', edgecolor='black')
ax1.set_title("Department vs. Total Employees")
ax1.set_xlabel("Departments")
ax1.set_ylabel("Number of Employees")
ax1.tick_params(axis='x', rotation=15)

for i, val in enumerate(hr_detail['Total_Employee']):
    ax1.text(hr_detail['Department'][i], val, str(val), ha='center', va='bottom')

ax2 = axes[0, 1]
ax2.bar(hr_detail['Department'],hr_detail['Total_Salary'], color='red', edgecolor='black')
ax2.set_title("Department vs. Total Salary")
ax2.set_xlabel("Departments")
ax2.set_ylabel("Total Salary Amount")
ax2.tick_params(axis='x', rotation=15)

for i, val in enumerate(hr_detail['Total_Salary']):
    ax2.text(hr_detail['Department'][i], val, str(val), ha='center', va='bottom')

ax3 = axes[1, 0]
ax3.bar(hr_detail['Department'],hr_detail['Total_Emp_Sat'], color='green', edgecolor='black')
ax3.set_title("Department vs. Total Employee Satisfaction")
ax3.set_xlabel("Departments")
ax3.set_ylabel("Number of Employees Satisfaction")
ax3.tick_params(axis='x', rotation=15)

for i, val in enumerate(hr_detail['Total_Emp_Sat']):
    ax3.text(hr_detail['Department'][i], val, str(val), ha='center', va='bottom')

ax4 = axes[1, 1]
ax4.bar(hr_detail['Department'],hr_detail['Total_Eng_Sur'], color='orange', edgecolor='black')
ax4.set_title("Department vs. Total Engagement Survey")
ax4.set_xlabel("Departments")
ax4.set_ylabel("Number of Engagement Survey")
ax4.tick_params(axis='x', rotation=15)

for i, val in enumerate(hr_detail['Total_Eng_Sur']):
    ax4.text(hr_detail['Department'][i], val, str(val), ha='center', va='bottom')
```
