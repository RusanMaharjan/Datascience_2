```python
# Aggregate Functions -> mean (average), sum, max, min, count
# KPI (Key Performance Indicator)
```

```python
import pandas as pd
```

```python
df = pd.read_csv('D:/Broadway/Data Science - 2/Employees.csv')
df.head()
```

```python
commission_mean = df['COMMISSION_PCT'].mean()
com_mean = "{:.2f}".format(commission_mean)
```

```python
df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(com_mean)
```

```python
mgr_mode = df['MANAGER_ID'].mode()
mgr_mode
```

```python
df['MANAGER_ID'] = df['MANAGER_ID'].fillna(mgr_mode)
```

```python
dep_mode = df['DEPARTMENT_ID'].mode()
dep_mode
```

```python
df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(50.0)
```

```python
df['DEPARTMENT_ID']
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
# Find average, max, min, total salary of employees
mean_salary = df['SALARY'].mean()
print(mean_salary)
```

```python
max_salary = df['SALARY'].max()
print(max_salary)
```

```python
min_salary = df['SALARY'].min()
print(min_salary)
```

```python
total_salary = df['SALARY'].sum()
print(total_salary)
```

```python
count_salary = df['SALARY'].count()
print(count_salary)
```

```python
# Find total number of employees in each department.
# df.groupby(column_name)[column_name].count()
emp_dep = df.groupby('DEPARTMENT_ID')['EMPLOYEE_ID'].count().reset_index()
emp_dep
```

```python
# Find total number of employees according to Job
job_emp = df.groupby('JOB_ID')['EMPLOYEE_ID'].count().reset_index()
job_emp
```

```python
# Find total salary of employees according to Job.
job_sal = df.groupby('JOB_ID')['SALARY'].sum().reset_index()
job_sal
```

```python
# Find total number of Job available in each Departments.
dep_job = df.groupby('DEPARTMENT_ID')['JOB_ID'].count().reset_index()
dep_job
```

```python
# Find average and total salary of employees by job
salary_details = df.groupby('JOB_ID').agg(
    average_salary = ('SALARY', 'mean'),
    total_salary = ('SALARY', 'sum'),
    max_salary = ('SALARY', 'max'),
    min_salary = ('SALARY', 'min')
).reset_index()
salary_details
```

```python
salary_details = df.groupby('JOB_ID')['SALARY'].agg(
    average_salary = ('mean'),
    total_salary = ('sum'),
    max_salary = ('max'),
    min_salary = ('min')
).reset_index()
salary_details
```
