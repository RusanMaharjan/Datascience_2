```python
from google import genai
```

```python
client = genai.Client()
```

```python
def gemini_llm(text):
    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=text,
    )
    return response.text
```

```python
content = input('Enter your prompt: ')
gemini_llm(content)
```
