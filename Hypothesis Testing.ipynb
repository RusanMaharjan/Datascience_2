# Hypothesis Testing

## Is the average total_price of all pizza sales different from $15.

```python
'''
1. Define alpha value (0.05) -> risk limit
2. Define Average Value / Average value column
3. Testing function
4. Define Null Hypothesis and Alternate Hypotheisis
5. Compare Alpha Value and P-Value for decision and conclusion.
'''
```

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('D:/Broadway/Data Science - 2/pizza_sales.csv')
df.head()
```

## One-Sample T-Test

```python
# Alpha value
alpha = 0.05
sample = 15 # sample data
total_price = df['total_price'] # Population data
```

```python
# testing one-sample t-test
t_statistics, pvalue = stats.ttest_1samp(total_price, popmean=sample)
```

```python
print(f'Ho: Total Average Price of Pizza Sales is equals to ${sample}.')
print(f'Ha: Total Average Price of Pizza Sales is not equals to ${sample}.')
print(f'Average Pizza Sales according to Data: ${total_price.mean():.2f}.')
print(f'Statistics: {t_statistics:.2f}')
print(f'Probability Value (P-Value): {pvalue}')
```

```python
if pvalue > alpha:
    print(f'Decision: Failed to reject null hypotheis.\nConclusion: Total Average Price of Pizza Sales is equals to ${sample}.')
else:
    print(f'Decision: Failed to accept null hypotheis.\nConclusion: Total Average Price of Pizza Sales is not equals to ${sample}.')
```

## Two-Sample T-Test

### Is the average unit_price of Classic pizzas vs. Chicken Pizzas equal?

```python
alpha = 0.05
```

```python
classic_price = df[df['pizza_category'] == 'Classic']['unit_price']
chicken_price = df[df['pizza_category'] == 'Chicken']['unit_price']
```

```python
t_statistics, pvalue = stats.ttest_ind(classic_price, chicken_price, equal_var=False)
```

```python
print('Ho: The average unit price of Classic Pizza and Chicken Pizza are Equal.')
print('Ha: The average unit price of Classic Pizza and Chicken Pizza are not Equal.')
print(f'Classic Pizza Average Price: ${classic_price.mean():,.2f}')
print(f'Chicken Pizza Average Price: ${chicken_price.mean():,.2f}')
```

```python
if pvalue > alpha:
    print('Decision: Failed to reject Null Hypothesis.\nConclusion: The average unit price of Classic Pizza and Chicken Pizza are Equal.')
else:
    print('Decision: Failed to accept Null Hypothesis.\nConclusion: The average unit price of Classic Pizza and Chicken Pizza are not Equal.')
```

## ANOVA Test (Analysis of Variance)

### The average total_price generated by all four pizza categories (Classic, Chicken, Supreme, and Veggie) differ significantly from one another?

```python
alpha = 0.05
```

```python
categories = df['pizza_category'].unique()
categories
```

```python
grouped_price = []
```

```python
for category in categories:
    pizza_price = df[df['pizza_category'] == category]['total_price']
    grouped_price.append(pizza_price)
```

```python
t_statistics, pvalue = stats.f_oneway(*grouped_price)
```

```python
mean_price = df.groupby('pizza_category')['total_price'].mean()
```

```python
print("Ho: The average total_price generated by all four pizza categories does not differ significantly from one another.")
print("Ha: The average total_price generated by all four pizza categories does differ significantly from one another.")
print('-'*36)
print(mean_price.map("${:,.2f}".format).to_markdown(numalign='left', stralign='left'))
print('-'*36)
```

```python
if pvalue > alpha:
    print("Decision: Failed to reject Null Hypothesis.\nConclusion: The average total_price generated by all four pizza categories does not differ significantly from one another.")
else:
    print("Decision: Failed to accept Null Hypothesis.\nConclusion: The average total_price generated by all four pizza categories does differ significantly from one another.")
```

## Chi-Squared Test

### Is pizza category and pizza size are independent or dependent to each other.

```python
alpha = 0.05
```

```python
contingency_table = pd.crosstab(df['pizza_category'], df['pizza_size'])
contingency_table
```

```python
# (number_of_rows - 1)*(number_of_columns - 1)
```

```python
t_statistis, pvalue, dof, ef = stats.chi2_contingency(contingency_table)
```

```python
print("Ho: Pizza Category and Pizza Size are dependent to each other.")
print("Ha: Pizza Category and Pizza Size are independent to each other.")
```

```python
if pvalue > alpha:
    print("Decision: Failed to reject Null Hypothesis.\nConclusion: Pizza Category and Pizza Size are dependent to each other.")
else:
    print("Decision: Failed to accept Null Hypotheis.\nCocnlusion: Pizza Category and Pizza Size are independent to each other.")
```
