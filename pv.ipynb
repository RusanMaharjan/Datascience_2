```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('D:/Broadway/Data Science - 2/HR_Dataset Refresh.csv')
```

```python
df.head()
```

```python
total_gender = df.groupby('Gender')['EmpID'].count().reset_index()
total_gender
```

```python
plt.figure(figsize=(6, 4))
plt.bar(total_gender['Gender'], total_gender['EmpID'], width=0.5)

for i, val in enumerate(total_gender['EmpID']):
    plt.text(total_gender['Gender'][i], val, str(val), va='bottom')
```

```python
ax = sns.barplot(x=total_gender['Gender'], y=total_gender['EmpID'])
for container in ax.containers:
   ax.bar_label(container)
```

```python
plt.hist(df['Salary'], edgecolor='black')
plt.xlabel('Salary')
plt.ylabel('Salary Distribution')
```

```python
mean_salary = df['Salary'].mean()
median_salary = df['Salary'].median()
```

```python
sns.histplot(df['Salary'], bins=10, kde='True', color='blue')
plt.axvline(mean_salary, color='red', label='Mean')
plt.axvline(median_salary, color='green', ls='--', label='Mean')
plt.legend()
```

```python
ax = sns.barplot(x=df['PerformanceScore'], y=df['Salary'], errorbar=None, 
            palette='viridis', hue=df['Gender'])

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')
```
