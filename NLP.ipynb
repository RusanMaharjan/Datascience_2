# Text PreProcessing

```python
import re # re -> Regular Expression
```

```python
text = 'Hello. My email is@   abc@info.com, def@info.com ghi@info.124  ## call at: 67345678 number 1234. Contact ## me.'
```

## Remove white space

```python
clean_text = re.sub(r'\s+',' ', text)
clean_text
```

## Remove unwanted text

```python
remove_has = re.sub(r'\s+', ' ', clean_text)
remove_has
```

```python
remove = re.sub(r'[@:##]', '', clean_text)
remove
```

```python
re_num = re.sub(r'number\s+\d+', '', remove)
re_num
```

```python
re_a = re.sub(r'@', '', remove_has)
re_a
```

```python
r = re.sub(r'\w+@\s+', '', remove_has)
r
```

# findall

```python
find_email = re.findall(r'\w+@\w+\.\w+', text)
find_email
```
